```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Age': [25, 30, np.nan, 40, 35],
    'Department': ['HR', 'Finance', 'Finance', np.nan, 'IT']
})

df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Department'] = df['Department'].fillna(df['Department'].mode()[0])

print(df)
```

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Age': [25, 30, np.nan, 40, 35],
    'Department': ['HR', 'Finance', 'Finance', np.nan, 'IT']
})

print("Original Dataset:")
print(df)

df_ffill = df.copy()
df_ffill.ffill(inplace=True)
print("Forward Fill:")
print(df_ffill)

df_bfill = df.copy()
df_bfill.bfill(inplace=True)
print("Backward Fill:")
print(df_bfill)
```

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Age": [25, 30, np.nan, 40, 35],
    "Department": ["HR", "Finance", np.nan, "Finance", "IT"]
})

print("Original Dataset:")
print(df)

df_drop_rows = df.dropna()
print("After dropping rows:")
print(df_drop_rows)

df_drop_cols = df.dropna(axis=1)
print("After dropping columns:")
print(df_drop_cols)
```

```python
import pandas as pd

df = pd.DataFrame({
    'ID': [1, 2, 2, 3, 4, 4],
    'Name': ['Alice', 'Bob', 'Bob', 'Charlie', 'David', 'David'],
    'Age': [25, 30, 30, 35, 40, 40]
})

print("Original Data:")
print(df)

df_exact = df.drop_duplicates()

print("After Removing Duplicates:")
print(df_exact)
```

```python
import pandas as pd

df = pd.DataFrame({
    'ID': [1, 2, 2, 3, 4, 4, 5],
    'Name': ['Alice', 'Bob', 'Bob', 'Charlie', 'David', 'David', 'David'],
    'Age': [25, 30, 30, 35, 40, 40, 40]
})

print("Original Data:")
print(df)

df_id = df.drop_duplicates(subset=['ID'])
print("Remove Duplicates Based on ID:")
print(df_id)

df_name = df.drop_duplicates(subset=['Name'])
print("Remove Duplicates Based on Name:")
print(df_name)
```

```python
import pandas as pd

df = pd.DataFrame({
    'Date': ['2025-01-05', '05/01/2025',
             'Jan 5, 2025', '2025.01.05']
})

print("Original Data:")
print(df)

df['Date'] = pd.to_datetime(
    df['Date'], errors='coerce'
).dt.strftime('%Y-%m-%d')

print(df)
```
